```python
import urllib.request

url = "https://ftp.ncbi.nlm.nih.gov/geo/series/GSE200nnn/GSE200167/suppl/GSE200167_Normalized_read_count_22Rv1_Ctrl_MYC-OE.txt.gz"

urllib.request.urlretrieve(
    url,
    "GSE200167_Normalized_read_count.txt.gz"
)

print("Dataset downloaded successfully!")
```

```python
import pandas as pd
```

```python
df = pd.read_csv(
    "GSE200167_Normalized_read_count.txt.gz",
    sep="\t",
    compression="gzip"
)
```

```python
myc = df[df["Unnamed: 0"].astype(str).str.upper() == "MYC"]

print(myc)
```

```python
df["Unnamed: 0"]
```

```python
if not myc.empty:
    print("MYC gene is present in the dataset.")
else:
    print("MYC gene is not present in the dataset.")
```

```python
print("MYC expression values:")
print(myc[["C1", "C2", "M1", "M2"]])
```

```python
df["Mean_Expression"] = df[["C1", "C2", "M1", "M2"]].mean(axis=1)
```

```python
max_index = df["Mean_Expression"].idxmax()

print("Gene with maximum expression:")
print(df.loc[max_index, "Unnamed: 0"])

print("Mean expression:")
print(df.loc[max_index, "Mean_Expression"])
```
