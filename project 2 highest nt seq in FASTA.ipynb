```python
from Bio.Seq import Seq
from Bio.SeqRecord import SeqRecord
from Bio import SeqIO
from collections import Counter
```

```python
fasta_text = """>Sample_Nucleotide_Sequence
ATGCGTACGTTAGCGATCGATGCTAGCTAGGCTAACGTTACGATCGATCGTACGATCGTAGCTAGCTAACG
"""

with open("sample.fasta", "w") as file:
    file.write(fasta_text)
```

```python
record = SeqIO.read("sample.fasta", "fasta")

sequence = str(record.seq).upper()

print("Sequence ID:", record.id)
print("Sequence:", sequence)
print("Sequence length:", len(sequence))
```

```python
nucleotide_count = Counter(sequence)

print("A =", nucleotide_count["A"])
print("T =", nucleotide_count["T"])
print("G =", nucleotide_count["G"])
print("C =", nucleotide_count["C"])
```

```python
highest_nucleotide = max(
    "ATGC",
    key=lambda nucleotide: nucleotide_count[nucleotide]
)

highest_count = nucleotide_count[highest_nucleotide]

print("Highest occurring nucleotide:", highest_nucleotide)
print("Count:", highest_count)
```

```python
length = len(sequence)

for nucleotide in "ATGC":
    percentage = (nucleotide_count[nucleotide] / length) * 100
    print(nucleotide, "=", nucleotide_count[nucleotide], 
          "(", round(percentage, 2), "%)")
```
